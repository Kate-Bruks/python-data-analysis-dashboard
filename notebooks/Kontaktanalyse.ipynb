```python
import pandas as pd
```

```python
contacts_raw = pd.read_excel('Contacts (Done).xlsx')

contacts = contacts_raw.copy()

print("--- Datei-Informationen Contacts ---")
print(contacts.info())
print("\n--- Fehlwerte in den Spalten ---")
print(contacts.isna().sum())
```

```python
# 1. ID in Zeichenkette umwandeln(Object)
contacts['Id'] = contacts['Id'].astype(str)

# 2.Datumsangaben in Datumsformat konvertieren
contacts['Created Time'] = pd.to_datetime(contacts['Created Time'], dayfirst=True)
contacts['Modified Time'] = pd.to_datetime(contacts['Modified Time'], dayfirst=True)

contacts['Contact Owner Name'] = contacts['Contact Owner Name'].astype('category')
print(contacts.info())
```

```python
contacts.to_csv('Contacts_Final.csv', index=False)
```
